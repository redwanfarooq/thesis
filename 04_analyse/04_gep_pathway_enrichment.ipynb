# Gene program pathway enrichment analyses

```python
import yaml
import pandas as pd
import gseapy as gp
```

# Load data

```python
cell_types = ["B", "DC", "Mono", "NK", "T"]
```

Load GEP spectra

```python
spectra = {}
for x in cell_types:
    with open(f"../results/tables/cite_seq/cohort_treatment_naive/gep/{x}/finalised/module_labels.yaml") as f:
        spectra[x] = pd.read_csv(
            f"../results/tables/cite_seq/cohort_treatment_naive/gep/{x}/finalised/gep_scores.tsv",
            sep="\t",
            header=0,
            index_col=0,
        ).rename(
            columns=yaml.safe_load(f)
        )
```

Load top 100 genes per GEP

```python
topgenes = {}
for x in cell_types:
    with open(f"../results/tables/cite_seq/cohort_treatment_naive/gep/{x}/finalised/module_labels.yaml") as f:
        topgenes[x] = pd.read_csv(
            f"../results/tables/cite_seq/cohort_treatment_naive/gep/{x}/finalised/topgenes.tsv",
            sep="\t",
            header=0,
    ).rename(
        columns=yaml.safe_load(f)
    )
```

# Gene Ontology

GMT file downloaded from https://www.gsea-msigdb.org/gsea/msigdb/download_file.jsp?filePath=/msigdb/release/2025.1.Hs/c5.go.v2025.1.Hs.symbols.gmt

```python
min_size = 20 # minimum gene set size
max_size = 200 # maximum gene set size

genesets = {
    x.split("\t")[0]: [gene.strip() for gene in x.split("\t")[2:]]
    for x in open("../data/ref/gene_sets/c5.go.v2025.1.Hs.symbols.gmt").readlines()
}
background = set(spectra[cell_types[0]].index) # set of all genes in the expression data
# Filter gene sets based on size and only keep genes present in the background set
genesets = {k: list(set(v).intersection(background)) for k, v in genesets.items() if min_size <= len(set(v).intersection(background)) <= max_size}
```

```python
enrichr = {}
for x in cell_types:
    print(f"Running GO enrichment analysis for GEPs from cell type: {x}")
    enrichr[x] = {
        gep: gp.enrichr(
            gene_list=topgenes[x][gep],
            gene_sets=genesets,
            background=background,
            outdir=None,
            cutoff=0.05,
            no_plot=True,
        )
        for gep in topgenes[x].columns.sort_values()
    }
```

```python
n_top = 25 # number of top enriched GO terms to extract per GEP

results = {}
for x in enrichr.keys():
    df_list = [enrichr[x][gep].results.sort_values("P-value").head(n_top).assign(Program=gep).rename(columns={"Term": "Gene Set"}) for gep in enrichr[x].keys()]
    results[x] = pd.concat(df_list).reset_index(drop=True)

# Save results
with pd.ExcelWriter("../results/tables/cite_seq/cohort_treatment_naive/gep/go_enrichment.xlsx") as writer:
    for x, df in results.items():
        df[["Program", "Gene Set", "Overlap", "P-value", "Genes"]].to_excel(writer, sheet_name=x, index=False)
```
